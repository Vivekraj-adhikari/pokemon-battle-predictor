```python
import requests

pokemon_data = requests.get("https://pokeapi.co/api/v2/pokemon?offset=0&limit=1024").json()
pokemon_list = pokemon_data["results"]
print(pokemon_list)
```

```python
pokemon_urls = [pokemon["url"] for pokemon in pokemon_list]
pokemons = []
for url in pokemon_urls:
    pokemon_details = requests.get(url).json()
    pokemon = pokemon_id = pokemon_details["id"]
    pokemon_name = pokemon_details["name"]
    pokemon_stats = {}
    for stats in pokemon_details["stats"]:
        stat_name = stats["stat"]["name"]
        stat_value = stats["base_stat"]
        pokemon_stats[stat_name] = stat_value
    pokemon_types = [pokemon_type["type"]["name"] for pokemon_type in pokemon_details["types"]]

    pokemon = {
        "id": pokemon_id,
        "name": pokemon_name,
        "stats": pokemon_stats,
        "types": pokemon_types
    }
    pokemons.append(pokemon)
```

```python
pokemon_flat = []
for pokemon in pokemons:
    pokemon_obj = {
        "id": pokemon["id"],
        "name": pokemon["name"],
        "hp": pokemon["stats"]["hp"],
        "attack": pokemon["stats"]["attack"],
        "defense": pokemon["stats"]["defense"],
        "special_attack": pokemon["stats"]["special-attack"],
        "special_defense": pokemon["stats"]["special-defense"],
        "speed": pokemon["stats"]["speed"],
        "type1": pokemon["types"][0],
        "type2": pokemon["types"][1] if len(pokemon["types"]) > 1 else "null"
    }
    pokemon_flat.append(pokemon_obj)
print(pokemon_flat)
```

```python
import pandas as pd
pokemon_df = pd.DataFrame(pokemon_flat)
pokemon_df.to_csv("pokemon.csv", index=False)
```

```python
import pandas as pd
import numpy as np

TYPES = [
    "Normal", "Fire", "Water", "Grass", "Electric",
    "Ice", "Fighting", "Poison", "Ground", "Flying",
    "Psychic", "Bug", "Rock", "Ghost", "Dragon",
    "Dark", "Steel", "Fairy"
]

type_chart_df = pd.DataFrame(
    np.ones((18, 18)),
    index = TYPES,
    columns = TYPES
)

type_chart_df.to_csv("type_chart.csv")
```

```python
import pandas as pd

type_chart = pd.read_csv("type_chart.csv", index_col=0)
print(type_chart.loc["Water", "Fire"])
```

```python
# Calculates type multipliers for a given attacking type against a defending type
def calculate_type_multiplier(attacking_type, defending_type1, defending_type2):
    if defending_type2 is not "null":
        return max(type_chart.loc[attacking_type, defending_type1], type_chart.loc[attacking_type, defending_type2])
    return type_chart.loc[attacking_type, defending_type1]
```
